```python
import os, sys, pytest

sys.dont_write_bytecode = True      # keeps __pycache__ out of your Git folder

nb_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
repo_root = "/".join(nb_path.split("/")[:-2])
if not repo_root.startswith("/Workspace"):
    repo_root = "/Workspace" + repo_root

os.chdir(repo_root)                 # so pytest.ini (pythonpath = src, testpaths = tests) is found

exit_code = pytest.main(["-v", "-p", "no:cacheprovider"])
assert exit_code == 0, f"Tests failed (pytest exit code {exit_code})"
```
