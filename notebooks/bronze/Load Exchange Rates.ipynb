```python
raw_df = spark.read.option("multiline",True).json(
    "/Volumes/retail_sales_dev/bronze/landing/exchange_rates/fx_rates_2010_2011.json"
)

raw_df.printSchema()
display(raw_df)
```

```python

from pyspark.sql.functions import current_timestamp, col

raw_df_with_meta = (raw_df
    .withColumn("_source_file", col("_metadata.file_path"))
    .withColumn("_ingested_at", current_timestamp())
)

query = (
    raw_df_with_meta.write
    .format("delta")
    .mode("overwrite")
    .option("mergeSchema", "true").saveAsTable("retail_sales_dev.bronze.exchange_rates_raw")
)
```
